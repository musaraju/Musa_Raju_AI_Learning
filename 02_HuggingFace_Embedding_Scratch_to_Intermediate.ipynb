```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
model=SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
print("Model loaded")
```

## What are embeddings?

Embeddings convert text into numerical vectors. Text with similar meaning often has vectors that are close in vector space. Sentence Transformers provides a convenient interface for this.

```python
sentences=["I love learning Python.","Python programming is interesting.","I enjoy playing cricket."]
embeddings=model.encode(sentences)
print("Number of sentences:",len(embeddings))
print("Embedding shape:",embeddings.shape)
print("First 10 values:",embeddings[0][:10])
```

```python
scores=cosine_similarity(embeddings)
for i in range(len(sentences)):
    for j in range(i+1,len(sentences)):
        print(f"{sentences[i]} <-> {sentences[j]}: {scores[i,j]:.4f}")
```

## Find the most similar sentence to a query

```python
query="I enjoy programming in Python."
q=model.encode([query])
q_scores=cosine_similarity(q,embeddings)[0]
ranking=sorted(zip(sentences,q_scores),key=lambda x:x[1],reverse=True)
for s,score in ranking: print(f"{score:.4f} | {s}")
print("Most similar:",ranking[0][0])
```

## Summary

We loaded a pretrained Hugging Face/Sentence Transformers model, generated embeddings, measured cosine similarity, and retrieved the closest sentence.
