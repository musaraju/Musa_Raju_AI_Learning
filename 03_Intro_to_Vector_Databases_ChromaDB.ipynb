## Key concepts

- **Collection:** group of records.
- **Document:** stored text.
- **ID:** unique record identifier.
- **Metadata:** additional information.
- **Embedding:** numerical representation used for semantic comparison.

Normal databases are strong at exact structured queries; vector databases are useful for semantic similarity search.

```python
import chromadb
client=chromadb.Client()
collection=client.get_or_create_collection(name="learning_notes")
documents=["Python is a popular programming language.","SQL is useful for working with databases.","Machine learning learns patterns from data.","Cricket is a popular sport."]
ids=["doc1","doc2","doc3","doc4"]
metadata=[{"topic":"python"},{"topic":"database"},{"topic":"machine learning"},{"topic":"sport"}]
collection.add(documents=documents,ids=ids,metadatas=metadata)
print("Documents stored:",collection.count())
```

```python
results=collection.query(query_texts=["programming and coding"],n_results=2)
for doc,dist in zip(results["documents"][0],results["distances"][0]): print(f"{doc} (distance: {dist:.4f})")
print("Metadata:",results["metadatas"][0])
```

## Conclusion

ChromaDB can store documents and metadata and retrieve relevant records using similarity search, making it useful for semantic search and RAG systems.
