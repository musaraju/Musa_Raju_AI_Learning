## Why chunking?

Large documents are divided into smaller pieces so each piece can be embedded and retrieved effectively. **Chunk size** controls piece length. **Overlap** repeats nearby text to reduce loss of context.

```python
def chunk_text(text,chunk_size=120,overlap=20):
    chunks=[]; start=0
    while start<len(text):
        end=min(start+chunk_size,len(text)); chunks.append(text[start:end])
        if end==len(text): break
        start=end-overlap
    return chunks

for size in [80,120,180]:
    chunks=chunk_text(document,size,20)
    print(f"\nChunk size {size}: {len(chunks)} chunks")
    for i,c in enumerate(chunks,1): print(f"{i}: {c}")
```

## Comparing chunk sizes

Smaller chunks are more focused but create more pieces. Larger chunks preserve more context but may include unrelated information. The best choice depends on the retrieval task.
