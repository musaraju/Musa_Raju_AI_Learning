```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import pandas as pd
sentences=["I love learning Python.","Python programming is interesting to me.","I enjoy playing cricket.","SQL is useful for data analysis.","I like studying machine learning."]
model=SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
embeddings=model.encode(sentences)
print("Embedding matrix shape:",embeddings.shape)
```

## Similarity matrix

```python
similarity_matrix=cosine_similarity(embeddings)
similarity_df=pd.DataFrame(similarity_matrix,index=range(1,6),columns=range(1,6))
similarity_df.round(3)
```

```python
upper=np.triu(similarity_matrix,k=1)
i,j=np.unravel_index(np.argmax(upper),upper.shape)
print("Most similar pair:")
print("1:",sentences[i])
print("2:",sentences[j])
print("Similarity score:",round(similarity_matrix[i,j],4))
```

## Query-based semantic search

Query: `I enjoy studying Python programming.`

```python
query="I enjoy studying Python programming."
q=model.encode([query])
scores=cosine_similarity(q,embeddings)[0]
ranking=sorted(zip(sentences,scores),key=lambda x:x[1],reverse=True)
for sentence,score in ranking: print(f"{score:.4f} | {sentence}")
best_sentence,best_score=ranking[0]
print("\nMost semantically similar sentence:")
print(best_sentence)
print("Similarity score:",round(best_score,4))
```

## Conclusion

A sentence can be converted into an embedding and compared with other sentences using cosine similarity. The highest-scoring result is the sentence whose meaning is closest to the query according to the pretrained model.
