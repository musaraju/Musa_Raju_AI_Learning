A word embedding represents a word as a numerical vector.

## Why numerical representations?

Computers can use vector mathematics to compare numerical representations of language. Similar contextual meaning can be reflected by similar vectors in real embedding models.

```python
import numpy as np
word_vectors={
    "python":np.array([0.90,0.80,0.20]),
    "programming":np.array([0.85,0.75,0.25]),
    "cricket":np.array([0.10,0.20,0.95]),
    "database":np.array([0.70,0.65,0.30])
}
```

```python
def cosine_similarity(a,b): return np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b))
for a,b in [("python","programming"),("python","cricket"),("programming","database")]:
    print(f"{a} <-> {b}: {cosine_similarity(word_vectors[a],word_vectors[b]):.4f}")
```

## Word embedding vs sentence embedding

**Word embedding:** represents an individual word/token.  
**Sentence embedding:** represents a whole sentence or passage as one vector.

Transformer models can also produce contextual representations, where meaning depends on surrounding words.

```python
print("Python vector:",word_vectors["python"])
print("Programming vector:",word_vectors["programming"])
```

## Conclusion

Embeddings turn language into numerical representations, and cosine similarity gives us a simple mathematical way to compare them.
