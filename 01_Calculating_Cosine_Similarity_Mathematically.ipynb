Cosine similarity measures how similar two vectors are by comparing their direction.

## Formula

`cosine similarity = (A · B) / (||A|| ||B||)`

The dot product is divided by the product of the vector magnitudes. Scores near 1 indicate similar direction, 0 indicates orthogonality, and -1 indicates opposite direction.

```python
import numpy as np

A=np.array([1,2,3]); B=np.array([2,4,6])
dot=np.dot(A,B)
mag_A=np.linalg.norm(A); mag_B=np.linalg.norm(B)
similarity=dot/(mag_A*mag_B)
print("Dot product:",dot)
print("Magnitude of A:",mag_A)
print("Magnitude of B:",mag_B)
print("Cosine similarity:",similarity)
```

## Step-by-step

For A=[1,2,3] and B=[2,4,6]:
1. A·B=(1×2)+(2×4)+(3×6)=28.
2. ||A||=√(1²+2²+3²).
3. ||B||=√(2²+4²+6²).
4. Divide the dot product by both magnitudes.

Because B is a positive multiple of A, the result is 1.

```python
def cosine_similarity(a,b):
    return np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b))

pairs=[(np.array([1,0]),np.array([1,0])),(np.array([1,0]),np.array([0,1])),(np.array([1,2]),np.array([2,1]))]
for i,(a,b) in enumerate(pairs,1): print(f"Pair {i}: {cosine_similarity(a,b):.4f}")
```

## Interpretation

The examples show identical, perpendicular, and moderately similar directions. Cosine similarity is commonly used to compare text embeddings.
